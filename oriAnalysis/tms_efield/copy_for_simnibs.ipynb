## Create a SIMNIBS_output folder for running SIMNIBS4.1
This copies data from brainsight sessions: only anatomicals and ROIs and also identifies subjects and compies their T2s from fs_subjects on data

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load meta-data from file
meta_file = '/d/DATD/datd/TMS_V3AB/analysis/V3AB_TMS_meta - Summary.csv'
summary_df = pd.read_csv(meta_file)

All_metadata = {row['Subject ID']: row for _, row in summary_df.iterrows()}
```

```python
brainsigh_path = '/d/DATD/datd/TMS_V3AB/brainsight'
simnibs_path = '/d/DATD/datd/TMS_V3AB/SIMNIBS_output'

# find the list of folders starting with 'sub' in the brainsight directory
subjects = [f for f in os.listdir(brainsigh_path) if os.path.isdir(os.path.join(brainsigh_path, f)) and f.startswith('sub')]
subjects.sort()

# Make corresponding directories in the SIMNIBS_output directory
for sub in subjects:
    if not os.path.exists(os.path.join(simnibs_path, sub)):
        os.makedirs(os.path.join(simnibs_path, sub))
    # For each subject copy all files except the files that end with .bsproj
    for f in os.listdir(os.path.join(brainsigh_path, sub)):
        if not f.endswith('.bsproj'):
            os.system('cp ' + os.path.join(brainsigh_path, sub, f) + ' ' + os.path.join(simnibs_path, sub))
    # Find the subject initials
    if sub == 'sub01':
        s_init = 'SSL'
    elif sub == 'sub13':
        s_init = 'AL0825'
    elif sub == 'sub15':
        s_init = 'K4S0421'
    elif sub == 'sub22':
        s_init = 'EMS0201'
    else:
        s_init = All_metadata[int(sub[-2:])]['Subject Initials']
    s_src_folder = os.path.join('/d/DATA/data/fs_subjects', s_init)+'src'
    # Find existing T2 files
    t2_files = [f for f in os.listdir(s_src_folder) if f.startswith('T2')]
    print(s_src_folder)
    print(t2_files)
    # Copy the first T2 file to the subject folder and name it as T2.nii.gz
    os.system('cp ' + os.path.join(s_src_folder, t2_files[0]) + ' ' + os.path.join(simnibs_path, sub, 'T2.nii'))
```
