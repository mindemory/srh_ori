## Run head segmentation using CHARM
Make sure that work_dir and subs to run are set up correctly and simply run the next cell to run the head-segmentation. It takes about 2:15 hours for each subject if both T1 and T2 are provided. The current code is optimized to run parallely for 4 subjects using 4 workers.

```python
import subprocess
import os
from concurrent.futures import ThreadPoolExecutor

def process_subject(sub):
    original_dir = os.getcwd()
    sub_id = f"sub{sub:02d}"
    work_dir = f"/d/DATD/datd/TMS_V3AB/SIMNIBS_output/{sub_id}/"
    os.chdir(work_dir)
    t2_exists = os.path.exists("T2.nii")

    if t2_exists:
        command = ["charm", f"{sub_id}", "T1.nii", "T2.nii", "--forceqform"]
    else:
        command = ["charm", f"{sub_id}", "T1.nii", "--forceqform"]

    print(f"Processing subject {sub_id}")

    try:
        print(command)
        result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
        print(f"Subject {sub_id} processing complete.")
        print("Standard Output:")
        print(result.stdout)
        print("Standard Error:")
        print(result.stderr)
        print(f"Return Code: {result.returncode}")
    except subprocess.CalledProcessError as e:
        print(f"Error executing command for subject {sub_id}: {e}")
    os.chdir(original_dir)

if __name__ == "__main__":
    #subs = list(range(1, 26))
    # For some reason 1 did not work out, check later
    # sub 3 T2.nii.gz issue
    subs = [23, 1]
    # Create a ThreadPoolExecutor with a specified number of worker threads
    num_threads = 1  
    with ThreadPoolExecutor(max_workers=num_threads) as executor:
        # Process each subject concurrently
        executor.map(process_subject, subs)
```
